```python
### A. REPRODUCIBLE NORMALIZATION PROBLEM
```

```python
import numpy as np
```

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
X_mean = np.mean(X)
X_mean
```

```python
X_std = np.std(X)
X_std
```

```python
X_normalized = (X - X_mean) / X_std
X_normalized
```

```python
X_normalized_mean = np.mean(X_normalized)
X_normalized_mean
```

```python
X_normalized_std = np.std(X_normalized)
X_normalized_std
```

```python
np.save("X_normalized.npy", X_normalized)
```

```python
#Required checks:
print("X Array:\n", X)
print("\nX_normalized:\n", X_normalized)
print("\nNormalized Mean (must be 0):", np.mean(X_normalized))
print("Normalized Std Dev (must be 1):", np.std(X_normalized))
```

```python
### B. CUBES DIVISIBLE BY 4 PROBLEM
```

```python
import numpy as np
```

```python
C = np.arange(1, 101)
C.resize(10, 10)
C
```

```python
C = np.power(C, 3)
C
```

```python
div_by_4 = C[C % 4 == 0]
div_by_4
```

```python
np.save("div_by_4.npy", div_by_4)
```

```python
#Required checks:
print("Shape of C:", C.shape)
print("\ndiv_by_4 array:\n", div_by_4)
print("\nNumber of selected elements:", len(div_by_4))
```

```python
### C. ABOVE-MEAN SQUARES PROBLEM
```

```python
import numpy as np
```

```python
S = np.arange(1, 37)
S.resize(6, 6)
S
```

```python
S = np.square(S)
S
```

```python
S_mean = np.mean(S)
S_mean
```

```python
above_mean = S[S > S_mean]
above_mean
```

```python
np.save("above_mean.npy", above_mean)
```

```python
#Required checks:
print("S Array:\n", S)
print("\nS_mean:", S_mean)
print("\nabove_mean array:\n", above_mean)
print("\nNumber of selected elements:", len(above_mean))
```
